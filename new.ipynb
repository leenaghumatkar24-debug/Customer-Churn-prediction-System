```python
#Exploring unique values in Categorical columns
features=["tenure","MonthlyCharges","TotalCharges"]
for col in df.columns:#it takes all columns from dataframe
    if col not in features: #if the above features are not present in this columns than print unique values of other columns 
        print(col,df[col].unique())
        print("--"*50)
```

```python
#Data Cleaning
```

```python
#Check values for mssing
df.isnull().sum()
```

```python
#Check for duplicate values
df.duplicated().sum()
```

```python
#drop duplicate values
df=df.drop_duplicates(keep="first")
```

```python
#After dropping duplicate values
df.duplicated().sum()
```

```python
#remove unecesary columns that are not useful for prediction
df=df.drop(['customerID'],axis=1)
```

```python
#After dropping unecessary column 
df 
```

```python
# Remove extra spaces from all categorical columns
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()
```

```python
#check which columns has blank spaces
for col in df.columns:
    print(col, df[col].astype(str).str.contains(" ").sum())
```

```python
#as we explore  are telco dataset the TotalCharges column has data in string/object format and blank spaces
```

```python
#Viewing TotalCharges Column
df["TotalCharges"].values #the below values are in string/object format
```

```python
df
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce") #errors="coerce" converts invalid columns into NAN
print(df["TotalCharges"].dtype)
```

```python
len(df[df["TotalCharges"]==" "]) #now there are 0 blank spaces in our dataset
```

```python
df.isnull().sum() #as we can see there sre total 11 missing values(NAN)i.e no value is stored
```

```python
#Fill the Missing values 
```

```python
df["TotalCharges"] = df["TotalCharges"].fillna(df["TotalCharges"].median())
```
